```python
def scrape_data(soup, season_text, league_name) :
    import csv 
    import os

    rows = soup.find(id="league-chemp").find_all('tr')

    base_dir = "./"  # Base directory where repositories will be created
    # Create repository directory if it doesn't exist
    repo_dir = os.path.join(base_dir, season_text)
    if not os.path.exists(repo_dir):
        os.makedirs(repo_dir)

    # Path to save the CSV file
    csv_file_path = os.path.join(repo_dir, league_name+".csv")

    # Open the CSV file in write mode
    with open(csv_file_path, 'w', newline='', encoding='utf-8') as csvfile:
        # Create a CSV writer object
        writer = csv.writer(csvfile)

        # Iterate over each table row
        for row in rows:
            # Extract data from each cell in the row
            row_data = [cell.get_text(strip=True) for cell in row.find_all(['td', 'th'])]

            # Write the row data to the CSV file
            writer.writerow(row_data)

    print(f"\tCSV file saved at: {csv_file_path}")
```

```python
driver.find_element(By.XPATH, "/html/body/div[1]/header/div/div[1]/div").click()                            ## find and click on league button
sleep(.5)

leagues = driver.find_elements(By.XPATH, "/html/body/div[1]/header/div/div[1]/ul/li")

for j in range(len(leagues)):
    
    league = driver.find_element(By.XPATH, f"/html/body/div[1]/header/div/div[1]/ul/li[{j+1}]")
    league_name = league.text
    print(f"scraping {league_name} league")
    league.click()
    sleep(2)
    
    
    driver.find_element(By.XPATH, "/html/body/div[1]/header/div/div[2]/div").click()
    sleep(0.5)
    seasons = driver.find_elements(By.XPATH, "/html/body/div[1]/header/div/div[2]/ul/li")
        
    for i in range(len(seasons)):
        
        season = driver.find_element(By.XPATH, f"/html/body/div[1]/header/div/div[2]/ul/li[{i+1}]")
        season_text = season.text.replace("/", "-")
        print(f"\tscraping {season_text} season")
        season.click()
        sleep(3)
        
        html = driver.page_source
        soup = BeautifulSoup(html, 'html.parser')
        
        scrape_data(soup, season_text, league_name)
        
        driver.find_element(By.XPATH, "/html/body/div[1]/header/div/div[2]/div").click()
        sleep(1)
        
    driver.find_element(By.XPATH, "/html/body/div[1]/header/div/div[1]/div").click()
    sleep(1)
```

```python
cat = ["How to join us?","My account","Need help ?","About us"]
import re
import time 
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By

def extract_text_with_formatting(soup):
    def extract_text_with_links(element):
        texts = []
        for item in element.contents:
            if item.name == 'br':
                texts.append('\n')
            elif item.name == 'a':
                texts.append(f"{item.text} [LINK: {item['href']}]")
            elif item.string:
                texts.append(item.string)
            elif item.name:
                texts.extend(extract_text_with_links(item))
        return texts

    text_content = extract_text_with_links(soup)
    formatted_text = ''.join(text_content)

    # Remove extra newlines
    formatted_text = re.sub(r'\n{3,}', '\n\n', formatted_text)
    return formatted_text

def parse_link(link):        
    print("parsing link: {}".format(link))    
    driver.get(link)
    time.sleep(1)
    html_content = driver.page_source
    soup = BeautifulSoup(html_content)
    question = soup.find("h2", class_="th-f-b th-fs-22 v20-th-fs-36").text
    faq_items = soup.find_all('section', {'data-type': 'faq-item'})

    content = extract_text_with_formatting(faq_items[0])
    res = "Question: "+question+"\nAnswer: "
    
    return res+content
```

```python
import os

directory = 'win'

def clean_and_update_files(directory):
    for filename in os.listdir(directory):
        if filename.endswith('.txt'):
            file_path = os.path.join(directory, filename)
            
            # Step 1: Clean the file by removing lines containing underscores
            with open(file_path, 'r') as file:
                lines = file.readlines()
            cleaned_lines = [line for line in lines if '_' not in line]
            with open(file_path, 'w') as file:
                file.writelines(cleaned_lines)
            
            # Step 2: Prepend the category and subcategory in the correct format
            parts = filename[:-4].split(')_')
            category = parts[0][1:].replace('_', ' ') + '?'
            subcategory = parts[1].replace('_', ' ')

            # Read the cleaned content of the file
            with open(file_path, 'r') as file:
                content = file.read()

            # Prepend the category and subcategory to the content
            new_content = f"Category: {category}\nSubcategory: {subcategory}\n\n{content}"

            # Write the new content back to the file
            with open(file_path, 'w') as file:
                file.write(new_content)

# Call the function with your directory
clean_and_update_files(directory)
```

```python
import os
from fpdf import FPDF

class PDF(FPDF):
    def header(self):
        # No header
        pass

    def footer(self):
        # No footer
        pass

def text_to_pdf(text_file, pdf_file):
    # Create instance of PDF class
    pdf = PDF()
    
    # Set margins
    pdf.set_margins(left=20, top=20, right=20)
    
    # Set auto page break
    pdf.set_auto_page_break(auto=True, margin=15)
    
    # Add a page
    pdf.add_page()
    
    # Set font - 'Helvetica' is a more readable font
    pdf.set_font("Helvetica", size=11)
    
    # Set text color to black
    pdf.set_text_color(0, 0, 0)
    
    # Open the text file in read mode
    with open(text_file, 'r', encoding='utf-8') as file:
        # Read the entire file content
        content = file.read()
        
        # Encode the content to handle special characters
        encoded_content = content.encode('latin-1', 'replace').decode('latin-1')
        
        # Add the content to the PDF
        pdf.multi_cell(0, 5, encoded_content)
    
    # Save the pdf with name .pdf
    pdf.output(pdf_file)

def convert_directory(input_dir, output_dir):
    # Create output directory if it doesn't exist
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    # Loop through all files in the input directory
    for filename in os.listdir(input_dir):
        if filename.endswith(".txt"):
            input_path = os.path.join(input_dir, filename)
            output_path = os.path.join(output_dir, filename[:-4] + ".pdf")
            text_to_pdf(input_path, output_path)
            print(f"Converted {filename} to PDF")

# Usage
input_directory = "files"
output_directory = "inwi_pdf"

convert_directory(input_directory, output_directory)
```
